# I. Tipos de Operadores

## Python incluye varios tipos de operadores:

-**Operadores aritméticos:** para realizar cálculos matemáticos

-**Operadores de comparación:** para comparar valores

-**Operadores lógicos:** para combinar condiciones

-**Operadores de asignación:** para asignar y modificar valores

-**Operadores de identidad:** para comparar la identidad de objetos

-**Operadores de pertenencia:** para verificar si un valor está en una secuencia


## 1. Operadores Aritméticos

| Operador | Significado | Ejemplo | Resultado |
|----------|-------------|---------|-----------|
| `+` | Suma | `5 + 3` | `8` |
| `-` | Resta | `5 - 3` | `2` |
| `*` | Multiplicación | `5 * 3` | `15` |
| `/` | División (flotante) | `5 / 2` | `2.5` |
| `//` | División entera | `5 // 2` | `2` |
| `%` | Módulo (residuo) | `5 % 2` | `1` |
| `**` | Potencia | `5 ** 3` | `125` |

````python
# Ejemplos de operadores aritméticos
a = 10
b = 3

print(f"Suma: {a + b}")
print(f"Resta: {a - b}")
print(f"Multiplicación: {a * b}")
print(f"División: {a / b}")
print(f"División entera: {a // b}")
print(f"Residuo: {a % b}")
print(f"Potencia: {a ** b}")
````

## 2. Operadores de Comparación

| Operador | Significado | Ejemplo | Resultado |
|----------|-------------|---------|-----------|
| `==` | Igual que | `5 == 3` | `False` |
| `!=` | Diferente de | `5 != 3` | `True` |
| `>` | Mayor que | `5 > 3` | `True` |
| `<` | Menor que | `5 < 3` | `False` |
| `>=` | Mayor o igual que | `5 >= 5` | `True` |
| `<=` | Menor o igual que | `5 <= 3` | `False` |

````python
# Ejemplos de operadores de comparación
a = int(input("Ingresa un número: "))
b = int(input("Ingresa otro número "))

print(f"¿a es igual a b? {a == b}")
print(f"¿a es diferente de b? {a != b}")
print(f"¿a es mayor que b? {a > b}")
print(f"¿a es menor que b? {a < b}")
print(f"¿a es mayor o igual que b? {a >= b}")
````

## 3. Operadores Lógicos

-Los operadores lógicos se utilizan principalmente para declaraciones condicionales.

-Permiten combinar múltiples condiciones y crear expresiones más complejas.

-Hay tres tipos de operadores lógicos: AND, OR y NOT.



| Operador | Descripción |
|----------|-------------|
| `and` |(Y) Devuelve TRUE cuando ambas expresiones son verdaderas |
| `or` | (O)Devuelve TRUE si al menos una condición es verdadera |
| `not` |(NO) Devuelve TRUE cuando la expresión no es verdadera |






### Tabla de Verdad para `and` (Y)
*Ambas condiciones deben ser `True` para que el resultado sea `True`.*

| A | B | A `and` B |
| :---: | :---: | :---: |
| `True` | `True` | `True` |
| `True` | `False` | `False` |
| `False` | `True` | `False` |
| `False` | `False` | `False` |

### Tabla de Verdad para `or` (O)
*Al menos una condición debe ser `True` para que el resultado sea `True`.*

| A | B | A `or` B |
| :---: | :---: | :---: |
| `True` | `True` | `True` |
| `True` | `False` | `True` |
| `False` | `True` | `True` |
| `False` | `False` | `False` |

### Tabla de Verdad para `not` (NO)
*Invierte el valor booleano.*

| A | `not` A |
| :---: | :---: |
| `True` | `False` |
| `False` | `True` |

````python
# Ejemplos de operadores lógicos
a = int(input("Ingresa un número: "))
b = int(input("Ingresa otro número "))
c = int(input("Ingresa otro número "))

print(f"¿a es mayor a b y a es mayor a c? {a > b and a > c}")
print(f"¿a es mayor a b o a es mayor a c? {a > b or a > c}")
print(f"¿a,b y c son numeros positivos? {a > 0 and b > 0 and c > 0}")
````

## 5. Operadores de Asignación

**Regla de oro:** Siempre se resuelve **lo de la derecha** primero, y el resultado se guarda en la **caja de la izquierda** (sobrescribiendo lo que había).

**(Para los ejemplos, asume que al inicio siempre tenemos:* `x = 10` *)*

| OPERADOR | NOMBRE | FUNCIÓN | EJEMPLO | RESULTADO |
| :---: | :--- | :--- | :--- | :---: |
| `=` | Asignación | Asigna a la variable del lado izquierdo cualquier variable o resultado del lado derecho | `x = 5` | `x = 5` |
| `+=` | Sumar y asignar | Realiza la suma y luego el resultado se asigna al operando de la izquierda | `x += 3` | `x = 13` |
| `-=` | Restar y asignar | Realiza la resta y luego el resultado se asigna al operando de la izquierda | `x -= 3` | `x = 7` |
| `*=` | Multiplicar y asignar | Realiza la multiplicación y luego el resultado se asigna al operando de la izquierda | `x *= 3` | `x = 30` |
| `/=` | Dividir y asignar | Realiza la división y luego el resultado se asigna al operando de la izquierda | `x /= 3` | `x = 3.333...` |
| `%=` | Módulo y asignar | Realiza el módulo y luego el resultado se asigna al operando de la izquierda | `x %= 3` | `x = 1` |
| `**=` | Exponente y asignar | Realiza exponente, y luego el resultado se asigna al operando de la izquierda | `x **= 2` | `x = 100` |
| `//=` | División entera y asignar | Realiza la división de piso y luego el resultado se asigna al operando de la izquierda | `x //= 3` | `x = 3` |

#II.  Estructuras de Control  🐍



### **¿Qué son estructuras de control?** 💻

Son instucciones que permiten controlar el flujo de un programa usando condicionales y bucles.
Dependiendo del resultado de las condiciones, se decide que acciones se van a ejecutar dentro del programa.
Existen diferentes tipos: secuenciales, condicionales e iterativas.



### 1.   Estructuras secuenciales:
Están compuestas por un número definido de acciones que se llevan a cabo en un orden específico y se ejecutan una tras otra.

**Pseudocódigo:**
```pseudocode
INICIO
    Leer variable1
    Leer variable2
    resultado = variable1 + variable2
    Mostrar resultado
FIN
```


````python
#Ejemplo en Python:
variable1 = int(input("Ingresa el primer número: "))
variable2 = int(input("Ingresa el segundo número: "))
resultado = variable1 + variable2
print(f"El resultado es: {resultado}")
````

### 2. Estructuras de control condicional:
 Son útiles en situaciones en la que puede haber una o más opciones disponibles para continuar y se debe tomar una decisión . Dependiendo de la respuesta a la pregunta, que siempre deberá ser VERDADERO (TRUE) o FALSO (FALSE), el algoritmo seguirá ciertas acciones e ignorará otras. Pueden ser simples, dobles y múltiples.

**2.1 Simples (IF):**
Se evalúa una condición, si el resultado es VERDADERO, se procede a ejecutar las acciones encerradas por esta estructura.

**Pseudocódigo:**
```pseudocode
INICIO
Si <CONDICIÓN> entonces
    Proceso
FIN
```

````python
#Ejemplo:
edad = int(input("Ingresa tu edad: "))
if edad >= 18:
    print("Eres mayor de edad")
````

**2.2 Dobles (IF, ELSE):**
Se añade una acción a ejecutarse en el caso de que la condición evaluada  devuelva el valor FALSO.

**Pseudocódigo:**
```pseudocode
INICIO
Si <CONDICIÓN> entonces
    Proceso 1
Si no
    Proceso 2
FIN
```

````python
#Ejemplo:
edad = int(input("Ingresa tu edad: "))
if edad >= 18:
    print("Eres mayor de edad")
else:
    print("Eres menor de edad")
````

### 2.3 Condicional Múltiple o Anidada (ELIF):

Permiten combinar varias estructuras condicionales para establecer controles más complejos sobre el flujo de procesos, representando una toma de decisión múltiple.

**Pseudocódigo:**
```pseudocode
INICIO
Si <CONDICIÓN 1> entonces
    Proceso 1
Si no
    Si <CONDICIÓN 2> entonces
        Proceso 2
    Si no
        Proceso 3
FIN

````python
#Ejemplo:
edad = int(input("Ingresa tu edad: "))
if edad >= 18:
    print("Eres mayor de edad")
else:
    if edad >= 13:
        print("Eres un adolescente")
    else:
        print("Eres un niño")
````

````python
#Ejemplo:
edad = int(input("Ingresa tu edad: "))
if edad >= 18:
    print("Eres mayor de edad")
elif edad >= 13:
    print("Eres un adolescente")
else:
    print("Eres un niño")
````

### 3. Estructuras de control iterativas

También se conocen como bucles (loops). Se utilizan cuando se necesita ejecutar repetidamente un determinado conjunto de acciones. El número de veces que se debe repetir dicha secuencia de acciones puede ser fijo o puede variar dependiendo de algún dato o condición a evaluar en el algoritmo.


### 3.1 Bucles con Número Fijo de Iteraciones (FOR)

Están controlados por un conteo. El algoritmo va contando la cantidad de repeticiones haciendo uso de una variable que recibe el nombre de **variable de iteración**, **índice** o **conteo**.

**Pseudocódigo:**
```pseudocode
INICIO
FOR <variable> DESDE <inicio> HASTA <fin> :
    Proceso
FIN
```
**Ejemplo - Mostrar números del 1 al 5:**
```pseudocode
FOR i DESDE 1 HASTA 5 :
    Mostrar i
FIN
```

````python
#La función range () se utiliza principalmente para generar y trabajar con una serie de números.
for i in range(1, 6):  # range(1,6) genera 1,2,3,4,5
    print(i)
````

````python
# Podemos personalizar range() para indicarle como avanzar en el intervalo,
#Ejemplo: Si queremos que avance de 2 en 2
for i in range(0, 10, 2):  # 0,2,4,6,8
    print(i)
````

````python
#Tambien se puede trabajar con una lista de elementos
#para ejecutar procesos para cada uno de los elementos en la lista
proteinas = ['hemoglobina','caseina','mioglobina','albumina']
for proteina in proteinas:
    print (proteina)
````

 **3.2 Con un número indeterminado de iteraciones (WHILE):**

Se uitiliza  para repetir un bloque de acciones  si no que el bloque de código se necesita ejecutar siempre que se cumpla una condición determinada (TRUE), y se deja de repetir cuando sea falsa (FALSE).

Necesitan un **contador** (o una condición de control) para evitar que se ejecute infinitamente.

**¿Cómo funciona el contador en un while?** 🤔

Debes seguir tres pasos:
1. Inicializar la variable antes de que empiece el bucle
2. Evaluar la variable en la condición del while.
3. Actualizar el valor de la variable dentro del bucle.

**Pseudocódigo:**
```pseudocode
i=0
WHILE <condición> HACER
    Proceso
    i = i + 1
FIN
```



````python
#Ejemplo:
i=1
while (i<6):
      print(f"El bucle se ha completado {i} veces")
      i=i+1
print("El bucle ha terminado")
````

### 3.3 Uso de Break y Continue

Existe una forma alternativa de interrumpir  los ciclos: utilizando la palabra reservada **break**.
Esto nos permite salir del ciclo incluso si la expresión evaluada en el bucle es verdadera. 😯

````python
#Ejemplo de break en un ciclo while:
variable = int(input("Introduzca un numero mayor a 5 : "))
while variable > 0:
   print ("Valor de la variable:", variable)
   variable = variable -1
   if variable == 5:
       break
````

````python
#Ejemplo de break en un ciclo for
numeros = [10, 20, 30, -5, 40, 50, 60]

print("Procesando números hasta encontrar uno negativo...")
for num in numeros:
    if num < 0:
        print(f"¡Número negativo {num} encontrado! ")
        break  # Detiene todo al encontrar negativo
    print(f"Procesando: {num}")
````

A diferencia de break, la función **continue ** NO interrumpe el ciclo, se salta a la siguente iteración



````python
#Ejemplo de continue con ciclo for
numeros = [10, 20, 30, -5, 40, 50, 60]

print("Procesando solo números positivos...")
for num in numeros:
    if num < 0:
        print(f"¡Número negativo {num} encontrado! ")
        continue  # Salta el negativo y continúa
    print(f"Procesando: {num}")
````

III. EJERCICIO INTEGRATIVO: Validador y Clasificador de Secuencias de ADN

## Instrucciones:

Escribe un programa en Python que haga lo siguiente:

1. **Solicitar** al usuario una secuencia de ADN (ejemplo: `"ATCG"`).
2. **Convertir** la secuencia a mayúsculas para evitar errores por minúsculas.
3. **Validar** que la secuencia SOLO contenga las letras: `A`, `T`, `C`, `G` o `N`.
   - Si encuentra **cualquier otro carácter** (ej: `X`, `Z`, `1`, etc.), el programa debe mostrar un mensaje de error.
4. **Si la secuencia es válida**, aplicar la siguiente lógica de clasificación:
   - Si la secuencia **contiene la letra "N"** (datos faltantes), imprimir:  
     `"Secuencia con datos faltantes"`.
   - Si **NO contiene "N"** y su **longitud es mayor a 10** bases, imprimir:  
     `"Secuencia valida y larga"`
   - Si **NO contiene "N"** y su **longitud es menor o igual a 10**, imprimir:  
     `"Secuencia valida pero corta."`


````python
# 1. Pedir la secuencia al usuario
secuencia = input("Introduce tu secuencia de nucleotidos (ej: ATCG): ")

# Convertimos a mayúsculas para evitar errores si el usuario escribe minusculas
secuencia = secuencia.upper()

# 2. Validar que SOLO tenga letras A, T, C, G o N
# Para esto, recorremos caracter por caracter y verificamos
es_valida = True  # Bandera (flag) que asumimos verdadera
for nuc in secuencia:
    if nuc not in "ATCGN":
        es_valida = False
        break  # Salimos del bucle en cuanto encontramos un error

# 3. Ahora aplicamos la lógica de clasificación usando condicionales
if not es_valida:
    print("Error: Caracter no identificado. Solo se permiten A, T, C, G o N.")
else:
    # Primero verificamos si tiene 'N'
    if "N" in secuencia:
        print("Secuencia con datos faltantes")
    else:
        # Si no tiene N, vemos la longitud
        if len(secuencia) > 10:
            print("Secuencia valida y larga")
        else:  # len(secuencia) <= 10
            print("Secuencia valida pero corta ")
````
