
## Estrategia 2: rellenar valores faltantes

A esta estrategia se le conoce como **imputación**,  es el procedimiento estadístico para reemplazar valores ausentes (NaN) por un valor estimado, permitiendo conservar las filas completas sin eliminar información.

La elección del valor sustituto depende de la naturaleza de la variable:

* **Moda:** Se utiliza para variables categóricas (texto o factores). Se asigna la categoría que aparece con mayor frecuencia en la columna.

* **Media:** Se utiliza para variables numéricas con distribución simétrica y sin valores extremos.

* **Mediana:** Se utiliza para variables numéricas con distribución asimétrica o con valores atípicos (outliers). Es robusta y no se deja sesgar por los extremos, ofreciendo un valor central más representativo.




```python
df_limpio = df.copy()

mediana_gc = df_limpio["GC_pct"].median()
mediana_cobertura = df_limpio["Cobertura_x"].median()
mediana_expresion = df_limpio["Expresion_relativa"].median()

df_limpio["GC_pct"] = df_limpio["GC_pct"].fillna(mediana_gc)
df_limpio["Cobertura_x"] = df_limpio["Cobertura_x"].fillna(mediana_cobertura)
df_limpio["Expresion_relativa"] = df_limpio["Expresion_relativa"].fillna(mediana_expresion)

df_limpio.isna().sum()
```


Para variables categóricas podemos utilizar la **moda** (el valor más frecuente).


```python
moda_tratamiento = df_limpio["Tratamiento"].mode()[0]
moda_sitio = df_limpio["Sitio"].mode()[0]

df_limpio["Tratamiento"] = df_limpio["Tratamiento"].fillna(moda_tratamiento)
df_limpio["Sitio"] = df_limpio["Sitio"].fillna(moda_sitio)

df_limpio.isna().sum()
```


> **Idea importante:** no existe una única manera correcta de tratar `NaN`.  
> La decisión depende de la pregunta, el origen del dato y el contexto experimental.


## Tabla dinámica con `pivot_table` (opcional)

Sirve para crear una tabla resumen reorganizando los datos, muy parecido a una tabla dinámica de Excel.


values="Expresion_relativa" indica qué dato queremos resumir;

index="Especie" coloca las especies en las filas;

columns="Tratamiento" coloca los tratamientos en las columnas;

y aggfunc="mean" indica que, si existen varias muestras para una misma especie y tratamiento, se calcule el promedio de la expresión relativa.

```python
tabla_resumen = df_limpio.pivot_table(
    values="Expresion_relativa",
    index="Especie",
    columns="Tratamiento",
    aggfunc="mean"
)

tabla_resumen.round(2)
```


## Solución sugerida


```python
reto = df.copy()

# 1. Faltantes
print(reto.isna().sum())

# 2. Rellenar variables numéricas
columnas_num = ["GC_pct", "Cobertura_x", "Expresion_relativa"]

for columna in columnas_num:
    reto[columna] = reto[columna].fillna(reto[columna].median())

# 3. Rellenar variables categóricas
for columna in ["Tratamiento", "Sitio"]:
    reto[columna] = reto[columna].fillna(reto[columna].mode()[0])

# 4. Eliminar duplicados
reto = reto.drop_duplicates()

# 5. Expresión media por tratamiento
print(
    reto.groupby("Tratamiento")["Expresion_relativa"]
        .mean()
        .round(2)
)

# 6. Top 5
display(
    reto.nlargest(5, "Expresion_relativa")[
        ["Muestra_ID", "Especie", "Tratamiento", "Expresion_relativa"]
    ]
)
```

```python
def nivel_expresion(valor):
    if valor < 2:
        return "Baja"
    elif valor < 4:
        return "Media"
    else:
        return "Alta"

reto["Nivel_expresion"] = reto["Expresion_relativa"].apply(nivel_expresion)

reto[["Muestra_ID", "Expresion_relativa", "Nivel_expresion"]].head()
```

```python
plt.figure(figsize=(8, 4))
sns.boxplot(
    data=reto,
    x="Tratamiento",
    y="Expresion_relativa"
)
plt.title("Expresión por tratamiento")
plt.show()
```

```python
plt.figure(figsize=(9, 5))
sns.scatterplot(
    data=reto,
    x="Cobertura_x",
    y="Expresion_relativa",
    hue="Nivel_expresion"
)
plt.title("Cobertura vs expresión")
plt.show()
```
