### 1.2 `print()` no es lo mismo que `return`

````python
def sumar_con_print(a, b):
    print(a + b)

def sumar_con_return(a, b):
    return a + b
````

````python
#función con print
resultado_1 = sumar_con_print(3, 4)
print("resultado_1 =", resultado_1)
````

````python
#función con return
resultado_2 = sumar_con_return(3, 4)
print("resultado_2 =", resultado_2)
````

¿Qué pasó?
<code>print()</code> solo muestra algo en pantalla. <code>return</code> entrega un valor al lugar donde la función fue llamada; por eso ese valor puede guardarse, combinarse o reutilizarse.



Una buena pregunta al escribir una función es: **¿quiero que la función muestre algo o que produzca un valor para seguir trabajando con él?**  return_funciones.png

### 1.3 Ejemplo aplicado: porcentaje de GC



````python
def porcentaje_gc(secuencia):
    cantidad_gc = secuencia.count("G") + secuencia.count("C")
    return cantidad_gc / len(secuencia) * 100
````

````python
secuencia_ejemplo = "ATGCGTACGCCTAG"
#llamamos la función y guardamos el resultado en la variable gc
gc = porcentaje_gc(secuencia_ejemplo)

print("% GC =", gc)
````

**Ventaja de las funciones:**

Una vez definida, la misma función puede aplicarse a cientos o miles de secuencias. Esa es una de las razones por las que aprender a diseñar funciones es tan útil en análisis reproducible.


### 1.4 Una función puede devolver más de un resultado

````python
def resumen_secuencia(secuencia):

    secuencia = secuencia.replace(" ", "") # quitamos espacio
    longitud = len(secuencia) # contar nucleótidos

    gc = secuencia.count("G") + secuencia.count("C")
    porcentaje = (gc / longitud * 100)

    return longitud, porcentaje
````

````python
longitud, gc = resumen_secuencia("ATG CCG TAA")

print("Longitud:", longitud)
print(f"GC: {gc:.1f}%")
````

## 3. Variables locales y globales

Una variable **global** se crea fuera de las funciones.  
Una variable **local** se crea dentro de una función y pertenece a esa llamada.

Vamos a observarlo.

````python
proyecto = "Proteómica"   # variable global

def describir_proyecto():
    mensaje = "Curso de python 2026..."   # variable local
    print("Proyecto:", proyecto)
    print(mensaje)

#llamar funciòn
describir_proyecto()
print("Fuera de la función todavía existe:", proyecto)
````

Ahora, ¿qué ocurrirá si intentamos usar `mensaje` fuera de la función?


Predice antes de ejecutar
¿Python imprimirá <code>"Analizando datos..."</code> o aparecerá un error?

````python
proyecto = "Proteómica"   # variable global

def describir_proyecto():
    #Dentro de la función
    mensaje = "Analizando datos..."   # variable local
    print("Proyecto:", proyecto)
    print(mensaje)

#Afuera de la función
print(mensaje)
````


El error no significa que la variable nunca existió; significa que <b>no existe en este ámbito</b>. <code>mensaje</code> era local en la función <code>describir_proyecto()</code>

## 4. Modificar variables globales desde una función

Para **leer** una variable global dentro de una función normalmente no necesitamos nada especial.  
Para **asignarle un nuevo valor** dentro de la función debemos declarar `global`.

```python
global nombre_variable
```

````python
secuencias_procesadas = 0

def registrar_secuencia():
    global secuencias_procesadas
    secuencias_procesadas += 1
    print("Procesando una secuencia...")

registrar_secuencia()
print("Total:", secuencias_procesadas)
registrar_secuencia()
print("Total:", secuencias_procesadas)
registrar_secuencia()
print("Total:", secuencias_procesadas)
````

 `global` **existe, pero úsalo con cuidado**. Muchas variables globales hacen que un programa sea más difícil de entender y depurar porque una función puede cambiar datos que están fuera de ella. En código científico suele ser más claro <b>recibir valores como parámetros y devolver resultados</b>.

### 4.1 Misma tarea, sin modificar una global

````python
def registrar_secuencia(contador):
    contador += 1
    #contador = contador + 1 #Tambien puedes usar esta forma de incremeto de valor
    return contador

# inicalizar variable
secuencias_procesadas = 0

#llamar función
secuencias_procesadas = registrar_secuencia(secuencias_procesadas)
print(secuencias_procesadas)
#llamamos función
secuencias_procesadas = registrar_secuencia(secuencias_procesadas)
print(secuencias_procesadas)
secuencias_procesadas = registrar_secuencia(secuencias_procesadas)

print("Total:", secuencias_procesadas)
````

## 5. Funciones recursivas

Una función es **recursiva** cuando se llama a sí misma.

Eso solo funciona bien si existen dos piezas:

1. **Caso base** → condición que detiene la recursión.
2. **Caso recursivo** → llamada con un problema más pequeño.


**Antes de ejecutar una función recursiva, pregúntate: <b>¿qué cambia en cada llamada y cómo llego al caso base?**

### 5.1 Primer ejemplo: cuenta regresiva

````python
def cuenta_regresiva(n):
    if n == 0:                 # caso base
        print("¡Despegue!")
        return

    print(n)
    cuenta_regresiva(n - 1)    # caso recursivo

cuenta_regresiva(5)
````

### Sigue las llamadas

Para `cuenta_regresiva(3)`:

```text
cuenta_regresiva(3)
└── cuenta_regresiva(2)
    └── cuenta_regresiva(1)
        └── cuenta_regresiva(0)  ← caso base
```


# **Conclusión Recursión**

La recursión es una técnica en la que una **función se llama a sí misma para resolver un problema dividiéndolo en versiones más pequeñas del mismo problema.** Toda función recursiva debe tener un **caso base**, que detiene las llamadas, y un caso recursivo, que acerca el problema al caso base.

**¿Cuándo usarla?**
Cuando el problema tiene una estructura naturalmente repetitiva o jerárquica, por ejemplo al recorrer árboles, carpetas, estructuras anidadas o algoritmos de búsqueda y división de problemas.

**¿Cuándo no usarla?**
Cuando existe una solución más simple y directa con ciclos, funciones ya disponibles en Python o métodos integrados. La recursión innecesaria puede hacer el código más difícil de entender y consumir más memoria.

## 6. ¿Qué pasa con la recursión infinita?

Considera:

```python
def repetir(n):
    return repetir(n)
```

La función vuelve a llamarse con **exactamente el mismo problema**, por lo que nunca se acerca a un caso base.

Python limita la profundidad de llamadas y finalmente produce `RecursionError`.

** Demostración segura para Colab**

 En la siguiente celda reduciremos <b>temporalmente</b> el límite de recursión y capturaremos el error. Después restauraremos el límite original. Así observamos el problema sin llenar la pantalla de mensajes.</div>
</div>

````python
import sys

limite_original = sys.getrecursionlimit()

def recursion_sin_salida():
    return recursion_sin_salida()

try:
    sys.setrecursionlimit(80)
    recursion_sin_salida()
except RecursionError:
    print(" Python detuvo la recursión: RecursionError")
finally:
    sys.setrecursionlimit(limite_original)

print("Límite restaurado a:", sys.getrecursionlimit())
````
